## News Automation. 
For a given stock ticker (get_ticker_given_name)
1. Get the latest News (get_yahoo_news)

```python
import yfinance as yf
```

```python
def get_news(ticker):
    import yfinance as yf

    print(f"\nFetching news for {ticker}...\n")

    stock = yf.Ticker(ticker)
    news = stock.news

    if not news:
        print("No news found")
        return

    for i, article in enumerate(news[:5], 1):
        content = article.get("content", {})

        title = content.get("title", "No title")
        description = content.get("description", "No description")

        print(f"{i}. {title}")
        print(f"   Description: {description[:200]}...\n")
```

```python
get_news("GOOG")
```
